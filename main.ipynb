```python
import tensorflow as tf
from tensorflow.keras.layers import (
    Input,
    GlobalAveragePooling2D,
    Dense,
    Dropout,
    Multiply,
    Concatenate
)
from tensorflow.keras.models import Model
from tensorflow.keras.applications import MobileNetV2
from vit_keras import vit

# Shared input
input_layer = Input(shape=(224, 224, 3))

# =========================
# MobileNetV2 Branch
# =========================
mobilenet = MobileNetV2(
    include_top=False,
    weights="imagenet",
    input_tensor=input_layer
)

mobilenet_features = GlobalAveragePooling2D()(
    mobilenet.output
)

# =========================
# ViT Branch
# =========================
vit_model = vit.vit_b16(
    image_size=224,
    pretrained=True,
    include_top=False,
    pretrained_top=False
)

vit_features = vit_model(input_layer)

# SE-style attention
attention = Dense(
    768,
    activation="sigmoid"
)(vit_features)

vit_features = Multiply()(
    [vit_features, attention]
)

# =========================
# Fusion
# =========================
fusion = Concatenate()(
    [mobilenet_features, vit_features]
)

# =========================
# Classifier
# =========================
x = Dense(512, activation="relu")(fusion)
x = Dropout(0.4)(x)

x = Dense(256, activation="relu")(x)
x = Dropout(0.3)(x)

output = Dense(1, activation="sigmoid")(x)

# =========================
# Final Model
# =========================
model = Model(
    inputs=input_layer,
    outputs=output
)

model.summary()
```

```python
import tensorflow as tf

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=1e-4
    ),
    loss="binary_crossentropy",
    metrics=[
        "accuracy",
        tf.keras.metrics.Precision(name="precision"),
        tf.keras.metrics.Recall(name="recall")
    ]
)
```

```python
import tensorflow as tf

callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor="val_loss",
        patience=10,
        restore_best_weights=True
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor="val_loss",
        factor=0.2,
        patience=5,
        min_lr=1e-7
    ),

    tf.keras.callbacks.ModelCheckpoint(
        filepath="womensafe_best.keras",
        monitor="val_accuracy",
        save_best_only=True,
        mode="max",
        verbose=1
    )
]
```

```python
history = model.fit(
    train_ds,
    validation_data=val_ds,
    epochs=50,
    callbacks=callbacks
)
```

```python
model.evaluate(test_ds)
```
